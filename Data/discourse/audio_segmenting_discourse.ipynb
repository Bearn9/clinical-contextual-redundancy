```python
import os
import shutil

ROOT_DIR = r"AUDIO/"

def move_mp3_files(root_dir):
    """
    For every folder under root_dir, moves all .mp3 files into a subfolder called 'mp3'.
    Existing 'mp3' folders are preserved (files are not overwritten).
    """
    for dirpath, dirnames, filenames in os.walk(root_dir):
        # Skip directories that are already 'mp3' to avoid recursion
        if os.path.basename(dirpath).lower() == "mp3":
            continue
        
        # Find all mp3 files in this folder
        mp3_files = [f for f in filenames if f.lower().endswith(".mp3")]
        if not mp3_files:
            continue  # nothing to move
        
        # Create mp3 subfolder if it doesn't exist
        mp3_folder = os.path.join(dirpath, "mp3")
        os.makedirs(mp3_folder, exist_ok=True)
        
        for mp3_file in mp3_files:
            src_path = os.path.join(dirpath, mp3_file)
            dest_path = os.path.join(mp3_folder, mp3_file)
            
            # Skip if file already exists in target folder
            if os.path.exists(dest_path):
                print(f"Skipping {src_path}, already exists in {mp3_folder}")
                continue
            
            shutil.move(src_path, dest_path)
            print(f"Moved {src_path} → {dest_path}")

# Run the function
move_mp3_files(ROOT_DIR)
```

```python
import os
import re
from pydub import AudioSegment
from datetime import datetime

RAW_AUDIO_DIR = r"AUDIO/TOPSY/1/6M/"
CHA_DIR = r"TRANSCRIPTS/TOPSY/TOPSY-1/6M/"
OUTPUT_DIR = r"AUDIO_CHUNKED/TOPSY/1/6M"
LOG_DIR = r"AUDIO_CHUNKED/Logs/TOPSY-1/6M"

MIN_CHUNK_DURATION = 5.0
PADDING = 0.1

os.makedirs(LOG_DIR, exist_ok=True)


# text normalization

def clean_text(text):
    text = re.sub(r"[^\w\s']", "", text)
    text = re.sub(r"\s+", " ", text)

    words = text.strip().split()
    cleaned = []

    for w in words:
        if w == "I":
            cleaned.append("I")
        else:
            cleaned.append(w.lower())

    return " ".join(cleaned)


# parsing cha files
def parse_cha_file(cha_path):
    segments = []

    with open(cha_path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    i = 0
    while i < len(lines):
        line = lines[i].strip()

        if line.startswith("*PAR"):
            ts_match = re.search(r"\x15(\d+)_(\d+)\x15", line)
            if ts_match:
                start_ms = int(ts_match.group(1))
                end_ms = int(ts_match.group(2))

                text = re.sub(r"\x15\d+_\d+\x15", "", line)
                text = re.sub(r"^\*PAR\d*:\s*", "", text).strip()

                words_text = ""
                if i + 1 < len(lines) and lines[i + 1].strip().startswith("%wor"):
                    words_text = lines[i + 1].strip()
                    i += 1

                segments.append(
                    (start_ms/1000.0, end_ms/1000.0, text, words_text)
                )

        i += 1

    return segments


def log_skip(log_file, start, end, duration, word_count, reason):
    log_file.write(
        f"SKIPPED | {start:.2f}-{end:.2f}s | "
        f"{duration:.2f}s | {word_count} words | {reason}\n"
    )


# main

def process_audio_file(audio_path, cha_path):

    base_name = os.path.basename(audio_path).replace(".wav", "")
    output_folder = os.path.join(OUTPUT_DIR, base_name)
    os.makedirs(output_folder, exist_ok=True)

    log_path = os.path.join(LOG_DIR, f"{base_name}_log.txt")
    log_file = open(log_path, "w", encoding="utf-8")

    log_file.write(f"Processing file: {base_name}\n")
    log_file.write(f"Timestamp: {datetime.now()}\n")
    log_file.write("="*60 + "\n")

    audio = AudioSegment.from_wav(audio_path)
    segments = parse_cha_file(cha_path)

    chunks = []
    i = 0

    while i < len(segments):

        start, end, text, words_text = segments[i]

        current_start = start
        current_end = end

        texts = [text]
        words_texts = [words_text] if words_text else []

        chunk_audio = audio[int(start*1000):int(end*1000)]

        j = i + 1

        while j < len(segments):

            next_start, next_end, next_text, next_words_text = segments[j]

            gap = next_start - current_end
            duration = current_end - current_start

            if duration >= 20.0:
                break

            if duration < 5.0:
                if gap >= 2.0:
                    break
            elif duration < 20.0:
                if gap >= 0.400:
                    break

            if gap < 0:
                gap = 0

            silence = AudioSegment.silent(duration=int(gap*1000))
            next_audio = audio[int(next_start*1000):int(next_end*1000)]

            chunk_audio += silence + next_audio

            current_end = next_end
            texts.append(next_text)

            if next_words_text:
                words_texts.append(next_words_text)

            j += 1

        final_duration = current_end - current_start
        num_words = sum(len(t.split()) for t in texts)

        reason = None

        if final_duration < MIN_CHUNK_DURATION:
            reason = "too_short_duration"
        elif num_words < 4:
            reason = "too_few_words"
        elif num_words > 60:
            reason = "too_many_words"

        if reason is None:

            chunk_id = len(chunks) + 1

            if PADDING > 0:
                chunk_audio += AudioSegment.silent(duration=int(PADDING*1000))

            wav_out = os.path.join(
                output_folder,
                f"{base_name}_{chunk_id:03d}.wav"
            )
            chunk_audio.export(wav_out, format="wav")

            merged_text = clean_text(" ".join(texts))

            txt_out = os.path.join(
                output_folder,
                f"{base_name}_{chunk_id:03d}.txt"
            )
            with open(txt_out, "w", encoding="utf-8") as f:
                f.write(merged_text)

            if any(words_texts):
                words_out = os.path.join(
                    output_folder,
                    f"{base_name}_{chunk_id:03d}_words.txt"
                )
                with open(words_out, "w", encoding="utf-8") as f:
                    f.write("\n".join(words_texts))

            chunks.append((chunk_id, current_start, current_end))

        else:
            log_skip(
                log_file,
                current_start,
                current_end,
                final_duration,
                num_words,
                reason
            )

        i = j

    log_file.write("="*60 + "\n")
    log_file.write(f"Total exported chunks: {len(chunks)}\n")
    log_file.close()

    print(f"Processed {base_name}: {len(chunks)} chunks")


for file in os.listdir(RAW_AUDIO_DIR):
    if file.endswith(".wav"):

        audio_path = os.path.join(RAW_AUDIO_DIR, file)
        cha_file = file.replace(".wav", ".cha")
        cha_path = os.path.join(CHA_DIR, cha_file)

        if os.path.exists(cha_path):
            process_audio_file(audio_path, cha_path)
        else:
            print(f"Skipping {file}, missing CHA file")
```
